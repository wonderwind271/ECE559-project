```python
import json
import ast
```

```python
seeds = list(range(3))
epochs_add = [100, 400, 500, 1000]
epochs_total = [100, 500, 1000, 2000, 4000, 10000]
ratios = [0.25,0.5,1,2,4]
margins = [.1,.5,1,1.5]
```

```python
import numpy as np
```

```python
all_info = {}
for ratio in ratios:
    filename = f'r{ratio}result.json'
    ratio_info = {}
    for e in epochs_total:
        ratio_info[e] = {'r': [], 'b': [], 'norm': []}
    with open(filename) as fp:
        data = json.load(fp)
        for margin in margins:
            for seed in seeds:
                # rs = []
                # bs = []
                # ns = []
                for i, epoch in enumerate(epochs_total):
                    plane = data[str(margin)][str(seed)][str(epoch)]
                    plane = ast.literal_eval(plane)
                    r = plane[2]/plane[1]
                    b = plane[0]
                    norm = np.linalg.norm(plane)
                    ratio_info[epoch]['r'].append(r)
                    ratio_info[epoch]['b'].append(b)
                    ratio_info[epoch]['norm'].append(norm)
    all_info[str(ratio)] = ratio_info

                
```

```python
import matplotlib.pyplot as plt
```

```python
def WMSE(exp, t):
    exp_array = np.array(exp)
    mse = np.mean((exp_array - t) ** 2)
    return mse/t
```

```python
rs = []
bs = []

for ratio in ratios:
    ratio_info = all_info[str(ratio)]
    er = [[] for i in range(6)]
    eb = [[] for i in range(6)]
    en = [[] for i in range(6)]
    for i, e in enumerate(epochs_total):
        er[i] += ratio_info[e]['r']
        eb[i] += ratio_info[e]['b']
        en[i] += ratio_info[e]['norm']
    rs.append(er)
    bs.append(eb)
    x_flat = []
    y_flat = []
    for xi, yi_list in zip(epochs_total, er):
        x_flat.extend([xi] * len(yi_list))
        y_flat.extend(yi_list)
    plt.scatter(x_flat, y_flat)
    plt.xlabel("epoch")
    plt.ylabel("r")
    plt.title(f"Scatter of overtime ratio for true ratio {ratio}")
    plt.savefig(f'ratio{ratio}_overtime.png')
    plt.clf()
```
